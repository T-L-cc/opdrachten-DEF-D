```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import random
import time
from scipy.optimize import curve_fit

# Dit is onze eigen library waar het imaging algoritme in staat
import imagingDEF2024 as imagingDEF
```

## Opdracht 3: Kalibratie sensor en foto

kalibratie-opstelling groep [12]
<img src="kalibratie foto.jpeg" width="400">

```python
afstand = np.array([5,10,15,20,25,30,35,40,44 ]) #cm
meting = np.array([141,282,447,595,755,898,1044,1186,1314]) #meting

test_meting = np.linspace(141, 1314,1000)

def testfit(x,a,b):
    return a*x + b

vals,covs = curve_fit(testfit,meting,afstand)

plt.figure()
plt.plot(meting,afstand, 'k.', label = 'metingen')
plt.plot(test_meting,testfit(test_meting,*vals), 'r--', label = 'curvefit')
plt.xlabel('gemeten bits')
plt.ylabel('afstand (cm)')
plt.legend()
plt.show()
print('Onze alpha is dus', vals[0], ', en onze beta', vals[1])
```
